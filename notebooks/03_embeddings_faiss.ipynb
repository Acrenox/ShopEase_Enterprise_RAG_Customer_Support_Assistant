# 3. Embeddings and FAISS index

Build the persisted FAISS index from the chunked support corpus using the configured embedding provider. The default `LLM_PROVIDER=local` keeps this workflow offline; Azure/Gemini require their respective credentials. Running this cell rebuilds `vectorstore/faiss_index`.

```python
from pathlib import Path
import os

ROOT = next(path for path in (Path.cwd(), *Path.cwd().parents) if (path / "ingestion").is_dir())
os.chdir(ROOT)

from app.core.config import settings
from app.rag.vectorstore import create_vectorstore
from ingestion.chunker import split_documents
from ingestion.document_loader import load_documents

documents = load_documents()
chunks = split_documents(documents)
vectorstore = create_vectorstore(chunks)

print(f"Embedding provider: {settings.llm_provider}")
print(f"FAISS vectors: {vectorstore.index.ntotal}")
print(f"Embedding dimensions: {vectorstore.index.d}")
print("Index saved to vectorstore/faiss_index")
```
