# 1. Offline ingestion

Load the registered support sources from `data/raw`, normalize their text, and inspect the metadata that will flow into chunking and retrieval. This uses the same loader as `python -m ingestion.build_index`.

```python
from pathlib import Path
import os

# Find the repository root so this cell works when launched from notebooks/ or the project root.
ROOT = next(path for path in (Path.cwd(), *Path.cwd().parents) if (path / "ingestion").is_dir())
os.chdir(ROOT)

from ingestion.document_loader import load_documents

documents = load_documents()
print(f"Loaded {len(documents)} documents")
for document in documents:
    print({
        "document_id": document.metadata["document_id"],
        "category": document.metadata["category"],
        "characters": len(document.page_content),
        "preview": document.page_content[:180].replace("\n", " "),
    })
```
