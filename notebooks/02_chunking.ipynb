# 2. Cleaned documents to chunks

Split loaded LangChain documents with the production chunker. Each chunk inherits source/category metadata and receives a stable-in-this-build `chunk_id` for inspection.

```python
from pathlib import Path
import os

ROOT = next(path for path in (Path.cwd(), *Path.cwd().parents) if (path / "ingestion").is_dir())
os.chdir(ROOT)

from ingestion.chunker import split_documents
from ingestion.document_loader import load_documents

documents = load_documents()
chunks = split_documents(documents)

print(f"Documents: {len(documents)} | Chunks: {len(chunks)}")
print(f"Chunk lengths: min={min(map(lambda chunk: len(chunk.page_content), chunks))}, "
      f"max={max(map(lambda chunk: len(chunk.page_content), chunks))}, "
      f"mean={sum(map(lambda chunk: len(chunk.page_content), chunks)) / len(chunks):.1f}")
for chunk in chunks[:5]:
    print({
        "chunk_id": chunk.metadata["chunk_id"],
        "source": chunk.metadata["source"],
        "category": chunk.metadata["category"],
        "characters": len(chunk.page_content),
        "preview": chunk.page_content[:140].replace("\n", " "),
    })
```
